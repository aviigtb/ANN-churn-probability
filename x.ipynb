```python
import tensorflow as tf
import tensorflow.keras.models as load_model
import numpy as np
import pandas as pd
import pickle
```

```python
model = load_model.load_model('Model.h5')
with open('one_hot_encoder_geo.pkl', 'rb') as f:
    one_hot_encoder_geo = pickle.load(f)

with open('label_encoder_gender.pkl', 'rb') as f:
    label_encoder_gender = pickle.load(f)

with open('scaler.pkl', 'rb') as f:
    scaler = pickle.load(f)
```

```python
input_data = {
    'CreditScore': 700,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
geo_encoded = one_hot_encoder_geo.transform([[input_data['Geography']]]).toarray()
geo_encoded_df = pd.DataFrame(geo_encoded, columns=one_hot_encoder_geo.get_feature_names_out(['Geography']))
```

```python
input_df = pd.DataFrame([input_data])
```

```python
input_df['Gender'] = label_encoder_gender.transform(input_df['Gender'])
input_df
```

```python
input_df = pd.concat([input_df.drop('Geography', axis=1), geo_encoded_df], axis=1)
```

```python
input_scaled = scaler.transform(input_df)
```

```python
pred = model.predict(input_scaled)
print("Predicted probability of churn:", pred[0][0])
```
